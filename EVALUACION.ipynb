```python

step3_base = pd.read_csv('STEP_2.csv')


cuantitativas = step3_base.select_dtypes(include=[np.number])
cualitativas = step3_base.drop(columns=cuantitativas.columns)
cuantitativas.columns
```

```python
#Realizamos diagrama de caja o bigote de cada columna del dataframe
fig = plt.figure(figsize=(15, 8))
cuantitativas.plot(kind='box', vert=False)
plt.title("Valores Atípicos del Dataframe (antes de sustitución) - Titanic")
plt.show()
```

```python
#Método aplicando Cuartiles. Encuentro cuartiles 0.25 y 0.75
y = cuantitativas

percentile25 = y.quantile(0.25) #Q1
percentile75 = y.quantile(0.75) #Q3
iqr = percentile75 - percentile25

limite_Superior_iqr = percentile75 + 1.5*iqr
limite_Inferior_iqr = percentile25 - 1.5*iqr
print("limite superior permitido", limite_Superior_iqr)
print("limite inferior permitido", limite_Inferior_iqr)
```

```python
#Obtenemos datos limpios del Dataframe
cuantitativas_sin_outliers = cuantitativas[(y<=limite_Superior_iqr)&(y>=limite_Inferior_iqr)]
cuantitativas_sin_outliers
```

```python
#Reemplazamos valores atípicos (nulos) del dataframe con "mediana"
#Realizamos una copia del dataframe
cuantitativas_limpio = cuantitativas_sin_outliers.copy()
cuantitativas_limpio= cuantitativas_limpio.fillna(round(cuantitativas_sin_outliers.median(), 1))
cuantitativas_limpio
```

```python

STEP_3 = pd.concat([cualitativas, cuantitativas_limpio], axis=1)
STEP_3
```

```python
#Corroboramos que no haya nulos
STEP_3.isnull().sum()
```

```python
#Convertimos el DataFrame resultante a CSV
STEP_3.to_csv("STEP_3.csv", index=False)
```
